##MISSING VALUE

###📝 I checked the dataset for missing or null values using df.isnull().sum().
###The analysis shows that all columns contain zero missing values,the dataset complete and ready for further analysis.

```python
#Handle missing values (impute or drop)
df.isnull().sum()
```

##drop

###📝The dropna() function was used to remove rows containing missing values from dataset .


```python
#DROP
df=df.dropna()
```

##Weight column

###📝The weight column was converted into a numeric format by removing the kg unit.


```python
df["Weight"]=df["Weight"].astype(str).str.replace("kg","",regex=False)
df["Weight"]=pd.to_numeric(df["Weight"],errors="coerce")
df["Weight"]=df["Weight"].fillna(df["Weight"].mean())
print(df.isnull().sum())
```

##Ram column

###📝The Ram column was converted into a numeric format by removing the GB unit.

```python
df["Ram"]=df["Ram"].astype(str).str.replace("GB","").str.strip().astype(float)
df["Ram"]=df["Ram"].fillna(df["Ram"].median())
print(df.isnull().sum())
```

##PRICE_EUROS

###📝The price_euros column are handled by replacing them with the mode of the column .the mode represent the most frequently occurring price in the dataset .no missing values remain price_euros column.

```python
df["Price_euros"]=df["Price_euros"].fillna(df["Price_euros"].mode()[0])
print(df.isnull().sum())
```

##DUPLICATES

###📝Duplicates records are checked to identify whether the dataset contains any repeated rows.in this dataset the duplicate check returned a value of 0.

```python
#Handle duplicates
df.duplicated().sum()
```

##COLUMN NAME RENAMEING

###📝Column name renameing is the process of changing existing column name to simpler and more meeaningful names.in this dataset the price_euros column is renamed to price, and the TypeName column is renamed Type.

```python
#column name rename

df.rename(columns={"Price_euros":"price","TypeName":"Type"},inplace=True)
print(df.rename)
```

##DROPPING UNWANTED COLUMNS

###📝Dropping unwanted columns is the process of removing columns that are not required for the analysis.in the dataset the product and screenResolution columns are removing because they are not needed for the further analysis.

```python
# dropping unwanted columns
df=df.drop(["Product","ScreenResolution"],axis=1)
print(df)
```

##Treat outliers

### 📝Outliers are data values that are unusually high(or) low compared with the other values in a dataset. in the analysis, the (IQR) interquartile range method is used to identify and remove outlier from price column .

```python
#Treat outliers if required
Q1=df["price"].quantile(0.25)
Q3=df["price"].quantile(0.75)
IQR=Q3-Q1
df=df[~((df["price"]<(Q1-1.5 * IQR))|(df["price"]>(Q3+1.5*IQR)))]
print(IQR)
```

##CHECK SKEWNESS

###📝Skewness is a statistical measure used to understand the distribution and numerical data. in this analysis the skewness of the price column is calculated using the skew() function.

```python
#Check skewness
print("skewness of price:",df["price"].skew())
```

##APPLYING LOG TRANSFORMATION


### 📝A log transformation is a mathematical transformation used to reduce he effect of large values and reduce the skewness of numerical data.

### np.log1p(df["price"])Applies the log transformation to the price values

```python
#apply transformations
import numpy as np
df["price_log"]=np.log1p(df["price"])
print(df["price_log"])
```

##DATATYPE

###📝Data type of values stored in each column of a dataset .in this step the data type of the RAM,WEIGHT,INCHES columns are checked using the datatypes function.

```python
#Convert data types if needed
print(df[["Ram","Weight","Inches"]].dtypes)
```

##FEATURE TRANSFORMATION


###📝Weight_per_inch is a derived feature that represents the weight of laptop relative to its screen size in inches. it is calculated by dividing the laptop weight by its screen size.


###📝Ram_category is categorical feature created by grouping RAM values into different categories based on the amount of RAM available.

```python
# New columns Feature transformations (date parts, derived fields if required for analysis)

df["Weight_per_Inch"]=df["Weight"] / df["Inches"]

df["Ram_category"]=pd.cut(df["Ram"],bins=[0,4,8,16,32,float("inf")],labels=["LOW","MEDIUM"," HIGH","VERY HIGH","EXTREME HIGH"])

print(df[["Ram","Weight","Inches","Weight_per_Inch","Ram_category"]].head())
```

##DISCOUNT PRICE

###📝Price represents the original cost while discounted price represents the cost after applying a 10% discount.

###📝EG:if price =1000,discounted price=900.


```python
# price and discounted_price
df["discounted_price"]=df["price"]*0.9
print(df["discounted_price"])
```

##RAM_GB

### 📝RAM_GB This code converts the values in the existing Ram column into floating point numbers (float) using the .astype() method and stores them in a new column called RAM_GB.

```python
#RAM_GB
df["RAM_GB"]=df["Ram"].astype(float)
print(df["RAM_GB"])
```

##SCREEN SIZE

###📝This code converts the values in the exiting inches column into floating point numbers(float) and saves them into a brand new column named screen_size.

```python
#SCREEN SIZE IN INCHES
df["Screen_size"]=df["Inches"].astype(float)
print(df["Screen_size"])
```

##WEIGHT IN KG

###📝 This converts the weight column values into decimal numbers(float) and stores them in a new column called WEIGHT_KG.


```python
#WEIGHT IN KG
df["Weight_kg"]=df["Weight"].astype(float)
print(df["Weight_kg"])
```

##PRICE IN INR

###📝This multiplies each values in the price column by 100 and creates new column named price_n_inr to store the calculated indian rupee amounts .



```python
#PRICE IN INR
df["price_inr"]=df["price"]*100
print(df["price_inr"])
```

##FINAL PRICE

###📝This calculates the  final price by subtracting the discounted_price from the price_in_inr,stores the result in a new column is a final_price.

```python
#FINAL PRICE
df["final_price"]=df["price_inr"]-df["discounted_price"]
print(df["final_price"])
```
